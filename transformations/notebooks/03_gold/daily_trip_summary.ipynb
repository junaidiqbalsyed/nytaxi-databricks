```python
from pyspark.sql.functions import col,  date_format, sum, mean, max, min, avg, count, col, round
```

```python
df = spark.read.table("nyctaxi.`02_silver`.yellow_trips_enriched")
```

```python
df = df.withColumn(
        'pickup_date', date_format('tpep_pickup_datetime', 'dd/MM/yyyy')
    )
```

```python
df = df.\
        groupBy(df['pickup_date']).\
        agg(
            count('*').alias('total_trips'),
            round(avg(col('passenger_count')),2).alias('avg_passengers_per_trip'),
            round(avg(col('trip_distance')),2).alias('avg_distance_per_trip'),
            round(avg(col('fare_amount')),2).alias('avg_fare_per_trip'),
            max(col('fare_amount')).alias('max_fare'),
            min(col('fare_amount')).alias('min_fare'),
            round(sum('fare_amount'),3).alias('total_revenue')
        )
```

```python
df.write.mode('overWrite').saveAsTable('nyctaxi.03_gold.daily_trip_summary')
```
