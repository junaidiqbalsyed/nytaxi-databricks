```python
from pyspark.sql.functions import current_date,lit
```

```python
schema = """
locationId int,
borough string,
zone string,
service_zone string
"""

df = spark.read.format('csv').load('/Volumes/nyctaxi/00_landing/data_sources/lookup/taxi_zone_lookup.csv', header = True, schema = schema)
```

```python
df = df.withColumns({
        'effective_date' : current_date(),
        'end_date' : lit(None)
    })
```

```python
df.write.mode('overwrite').saveAsTable('nyctaxi.02_silver.taxi_zone_lookup')
```
