```python
import warnings
warnings.filterwarnings('ignore')
import shutil
import os
import urllib.request as request
```

```python
landing_dir_path = f"/Volumes/nyctaxi/00_landing/data_sources/lookup"
os.makedirs(landing_dir_path,  exist_ok=True)

lookup_url = "https://d37ci6vzurychx.cloudfront.net/misc/taxi_zone_lookup.csv"

response = request.urlopen(lookup_url)

file_path = os.path.join(landing_dir_path, 'taxi_zone_lookup.csv')

with open(file_path, 'wb') as f:
    shutil.copyfileobj(response, f)
```
